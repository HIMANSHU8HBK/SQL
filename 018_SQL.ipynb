```python
%%sql

CREATE TABLE CustomersArchive AS
SELECT * FROM Customers WHERE 1=0;  -- Create empty clone table
```

```python
%%sql

SELECT * FROM CustomersArchive;
```

```python
%%sql

INSERT INTO CustomersArchive
SELECT * FROM Customers 
WHERE Country IN ('Germany', 'USA');
```
